````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# [第 2 格] 建立資料庫與初始資料
# 每次執行時先刪除舊表，再建立活動表 (events) 與報名紀錄表 (registrations) 並寫入初始資料

def init_db():
    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()

        # 刪除已存在的表格以確保每次執行都重新初始化
        cursor.execute("DROP TABLE IF EXISTS registrations;")
        cursor.execute("DROP TABLE IF EXISTS events;")

        # 建立活動資料表
        cursor.execute("""
        CREATE TABLE events (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT UNIQUE NOT NULL,
            event_date TEXT NOT NULL,
            capacity INTEGER NOT NULL
        );
        """)

        # 建立報名紀錄資料表
        cursor.execute("""
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            event_id INTEGER NOT NULL,
            people_count INTEGER NOT NULL,
            status TEXT NOT NULL CHECK(status IN ('active', 'cancelled')),
            FOREIGN KEY (event_id) REFERENCES events(id)
        );
        """)

        # 寫入題意要求的 3 場初始活動資料
        initial_events = [
            ("迎新晚會", "2026-11-20", 6),
            ("程式工作坊", "2026-11-25", 4),
            ("籃球友誼賽", "2026-11-28", 3)
        ]
        # 使用 ? 參數防止 SQL 注入
        cursor.executemany("INSERT INTO events (name, event_date, capacity) VALUES (?, ?, ?);", initial_events)
        conn.commit()
        print("✅ 資料庫與初始資料建立完成！")
    finally:
        conn.close()

# 執行初始化
init_db()
````

````python
# [第 3 格] 顯示資料表內容
# 使用 pandas 讀取 SQLite 資料庫，並用 display() 漂亮地呈現目前的表格資料

def show_tables():
    conn = sqlite3.connect("activity_system.db")
    try:
        print("=== 📌 活動資料表 (events) ===")
        df_events = pd.read_sql_query("SELECT * FROM events;", conn)
        display(df_events)

        print("\n=== 📌 報名紀錄資料表 (registrations) ===")
        df_regs = pd.read_sql_query("SELECT * FROM registrations;", conn)
        display(df_regs)
    finally:
        conn.close()

# 顯示當前資料表內容
show_tables()
````

````python
# [第 4 格] AI 解讀函式
# 呼叫 Groq API 解析語意，並使用快取 (cache) 與驗證邏輯（允許清單 + 正規表示式）來確保解析品質與穩定度

# 使用 dict 暫存 AI 解析結果，避免同一句話重複呼叫 API
ai_cache = {}

# 定義允許的清單 (White lists)
ALLOWED_ACTIONS = ["報名", "取消", "查詢名額", "統計查詢"]
ALLOWED_EVENTS = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

def parse_with_ai(user_input: str):
    # 先檢查快取是否有這句話的解讀紀錄
    if user_input in ai_cache:
        return ai_cache[user_input]

    prompt = f"""
你是一個活動報名系統的助理，請從使用者的句子中抽取「動作」、「活動」與「人數」。
- 動作只有：報名、取消、查詢名額、統計查詢（問各活動狀況時）
- 活動只有：迎新晚會、程式工作坊、籃球友誼賽
- 人數：包含攜伴，若提及「我和兩個朋友」請計算總人數為 3。未提及請回覆 1。

使用者說："{user_input}"
請簡短回覆即可。
"""

    raw_response = ""
    # 限流重試機制：最多重試 3 次，遇 429 錯誤等待 30 秒
    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_response = response.choices[0].message.content.strip()
            # 每次呼叫後固定等待 2 秒以符合 Rate Limit 規範
            time.sleep(2)
            break
        except Exception as e:
            if "429" in str(e):
                print(f"⚠️ 觸發 API 限流 (429)，等待 30 秒後重試 (第 {attempt + 1}/3 次)...")
                time.sleep(30)
            else:
                raw_response = f"API 發生錯誤: {e}"
                break

    # --- 程式比對與驗證 (不直接盲目相信 AI) ---

    # 1. 動作比對 (允許清單)
    parsed_action = "未知"
    for act in ALLOWED_ACTIONS:
        if act in raw_response:
            parsed_action = act
            break

    # 2. 活動比對 (允許清單)
    parsed_event = "未知"
    for ev in ALLOWED_EVENTS:
        if ev in raw_response:
            parsed_event = ev
            break

    # 3. 人數抽取 (使用正規表示式找出數字，找不到預設為 1)
    # 搜尋 "人數: 3" 或 "人數 3" 或獨立數字
    numbers = re.findall(r'\d+', raw_response)
    if numbers:
        parsed_count = int(numbers[0])
    else:
        parsed_count = 1

    # 封裝結果
    result = {
        "raw": raw_response,
        "action": parsed_action,
        "event": parsed_event,
        "count": parsed_count
    }

    # 存入快取並回覆
    ai_cache[user_input] = result
    return result
````

````python
# [第 5 格] AI 解讀測試
# 測試 9 句（包含正常、錯誤、邊界），測試格式為 (使用者說的話, 預期動作, 預期對象, 預期數量)，"*" 代表不在意

test_cases = [
    ("我想報名迎新晚會", "報名", "迎新晚會", 1),
    ("我和兩個朋友要參加程式工作坊", "報名", "程式工作坊", 3),
    ("幫我取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("請問程式工作坊還有名額嗎", "查詢名額", "程式工作坊", 1),
    ("想知道各活動報名狀況", "統計查詢", "*", "*"),
    ("帶4個朋友報名迎新晚會", "報名", "迎新晚會", 5),
    ("今天天氣真好想去散步", "未知", "未知", "*"),
    ("幫我報名神秘演唱會", "未知", "未知", "*"),
    ("幫我取消程式工作坊", "取消", "程式工作坊", 1)
]

pass_count = 0

print("==========================================")
print("🧪 AI 解讀測試開始 (測試數量 9 句)")
print("==========================================")

for idx, (text, exp_act, exp_evt, exp_cnt) in enumerate(test_cases, 1):
    res = parse_with_ai(text)

    # 驗證動作
    act_ok = (exp_act == "*") or (res["action"] == exp_act)
    # 驗證活動
    evt_ok = (exp_evt == "*") or (res["event"] == exp_evt)
    # 驗證數量
    cnt_ok = (exp_cnt == "*") or (res["count"] == exp_cnt)

    is_passed = act_ok and evt_ok and cnt_ok
    if is_passed:
        pass_count += 1

    status = "✅ PASS" if is_passed else "❌ FAIL"

    print(f"#{idx:02d} [{status}] 使用者說: \"{text}\"")
    print(f"  ├─ AI 原始回覆: {res['raw']}")
    print(f"  └─ 系統解讀結果: 動作={res['action']}, 活動={res['event']}, 人數={res['count']}\n")

print("==========================================")
print(f"🎯 測試結果：共 {len(test_cases)} 句，通過 {pass_count} 句。")
print("==========================================")
````

````python
# [第 6 格] 功能函式 (報名、取消、查詢名額、統計查詢)
# 包含所有商業邏輯判斷 (A1 人數限制 1~3、A2 個人上限 2 場、A3 統計 JOIN)

def get_available_events_text():
    """輔助函式：列出目前所有開放的活動"""
    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT name, event_date, capacity FROM events;")
        events = cursor.fetchall()
        msg = "⚠️ 無法判斷您的意圖或活動名稱。目前開放的活動清單如下：\n"
        for ev in events:
            msg += f"• {ev[0]}（日期：{ev[1]}，總名額：{ev[2]} 人）\n"
        return msg
    finally:
        conn.close()

def do_statistics():
    """A3 進階需求：使用 LEFT JOIN 與 GROUP BY 列出每個活動統計資料"""
    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()
        query = """
        SELECT
            e.name,
            e.event_date,
            e.capacity,
            COALESCE(SUM(CASE WHEN r.status = 'active' THEN r.people_count ELSE 0 END), 0) AS registered,
            e.capacity - COALESCE(SUM(CASE WHEN r.status = 'active' THEN r.people_count ELSE 0 END), 0) AS remaining
        FROM events e
        LEFT JOIN registrations r ON e.id = r.event_id
        GROUP BY e.id;
        """
        cursor.execute(query)
        rows = cursor.fetchall()

        output = "📊 【各活動報名統計狀況】\n"
        output += f"{'活動名稱':<10} | {'日期':<10} | {'總名額':<6} | {'已報名':<6} | {'剩餘名額':<6}\n"
        output += "-" * 55 + "\n"
        for row in rows:
            output += f"{row[0]:<10} | {row[1]:<10} | {row[2]:<8} | {row[3]:<8} | {row[4]:<8}\n"
        return output
    finally:
        conn.close()

def handle_registration(student_name: str, event_name: str, people_count: int):
    """報名邏輯判斷 (含 A1、A2 邏輯)"""
    if not student_name or not student_name.strip():
        return "❌ 報名失敗：必須提供學生姓名。"

    # A1 需求：人數不在 1~3 之間時由程式拒絕
    if people_count < 1 or people_count > 3:
        return f"❌ 報名失敗：單次報名人數必須介於 1 到 3 人之間（您提出的申請人數為 {people_count} 人）。"

    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()

        # 1. 取得活動 ID 與容量
        cursor.execute("SELECT id, capacity FROM events WHERE name = ?;", (event_name,))
        event_row = cursor.fetchone()
        if event_row is None:
            return f"❌ 報名失敗：找不到活動「{event_name}」。"

        event_id, capacity = event_row

        # 2. 檢查重複報名：同一位學生不可重複報名同一個活動 (status='active')
        cursor.execute("""
        SELECT id FROM registrations
        WHERE student_name = ? AND event_id = ? AND status = 'active';
        """, (student_name, event_name_to_id(cursor, event_name)))
        if cursor.fetchone() is not None:
            return f"❌ 報名失敗：【{student_name}】已經重複報名「{event_name}」。"

        # 3. A2 需求：每位學生最多同時報名 2 個活動 (status='active')
        cursor.execute("""
        SELECT COUNT(DISTINCT event_id) FROM registrations
        WHERE student_name = ? AND status = 'active';
        """, (student_name,))
        active_count_row = cursor.fetchone()
        active_count = active_count_row[0] if active_count_row else 0

        if active_count >= 2:
            return f"❌ 報名失敗：【{student_name}】已同時報名 {active_count} 個活動，達到每人上限 2 個活動的限制！"

        # 4. 計算目前剩餘名額
        cursor.execute("""
        SELECT COALESCE(SUM(people_count), 0) FROM registrations
        WHERE event_id = ? AND status = 'active';
        """, (event_id,))
        sum_row = cursor.fetchone()
        current_registered = sum_row[0] if sum_row else 0
        remaining = capacity - current_registered

        if people_count > remaining:
            return f"❌ 報名失敗：活動「{event_name}」剩餘名額僅 {remaining} 人，無法滿足您的 {people_count} 人申請。"

        # 5. 寫入報名紀錄
        cursor.execute("""
        INSERT INTO registrations (student_name, event_id, people_count, status)
        VALUES (?, ?, ?, 'active');
        """, (student_name, event_id, people_count))
        conn.commit()

        return f"✅ 報名成功！【{student_name}】已成功報名「{event_name}」{people_count} 位。"
    finally:
        conn.close()

def handle_cancel(student_name: str, event_name: str):
    """取消報名邏輯"""
    if not student_name or not student_name.strip():
        return "❌ 取消失敗：必須提供學生姓名。"

    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()

        # 尋找 active 狀態的報名紀錄
        cursor.execute("""
        SELECT r.id, r.people_count
        FROM registrations r
        JOIN events e ON r.event_id = e.id
        WHERE r.student_name = ? AND e.name = ? AND r.status = 'active';
        """, (student_name, event_name))

        reg_row = cursor.fetchone()
        if reg_row is None:
            return f"❌ 取消失敗：查無【{student_name}】對「{event_name}」的有效報名紀錄。"

        reg_id, count = reg_row
        cursor.execute("UPDATE registrations SET status = 'cancelled' WHERE id = ?;", (reg_id,))
        conn.commit()
        return f"✅ 取消成功！【{student_name}】對「{event_name}」的報名已取消，釋出 {count} 個名額。"
    finally:
        conn.close()

def handle_query_capacity(event_name: str):
    """查詢剩餘名額邏輯"""
    conn = sqlite3.connect("activity_system.db")
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT id, capacity FROM events WHERE name = ?;", (event_name,))
        event_row = cursor.fetchone()
        if event_row is None:
            return f"❌ 查詢失敗：找不到活動「{event_name}」。"

        event_id, capacity = event_row

        cursor.execute("""
        SELECT COALESCE(SUM(people_count), 0) FROM registrations
        WHERE event_id = ? AND status = 'active';
        """, (event_id,))
        sum_row = cursor.fetchone()
        registered = sum_row[0] if sum_row else 0
        remaining = capacity - registered

        return f"ℹ️ 活動「{event_name}」總名額：{capacity} 人，已報名：{registered} 人，剩餘名額：{remaining} 人。"
    finally:
        conn.close()

def event_name_to_id(cursor, event_name):
    """輔助函式：用活動名稱找 ID"""
    cursor.execute("SELECT id FROM events WHERE name = ?;", (event_name,))
    row = cursor.fetchone()
    return row[0] if row else None
````

````python
# [第 7 格] 主流程處理器與流程測試 (9 步依序執行，最後印出資料表)

def process_user_request(student_name: str, user_input: str) -> str:
    """系統核心主流程"""
    if not user_input or not user_input.strip():
        return "請輸入您的需求。"

    # 1. AI 解析語意
    parsed = parse_with_ai(user_input)
    action = parsed["action"]
    event_name = parsed["event"]
    people_count = parsed["count"]

    # 2. 依動作分流
    if action == "統計查詢":
        return do_statistics()
    elif action == "未知" or event_name == "未知":
        return get_available_events_text()
    elif action == "查詢名額":
        return handle_query_capacity(event_name)
    elif action == "報名":
        return handle_registration(student_name, event_name, people_count)
    elif action == "取消":
        return handle_cancel(student_name, event_name)
    else:
        return get_available_events_text()

# --- 重置資料庫並執行 9 步流程測試 ---
init_db()

flow_steps = [
    ("王小明", "報名迎新晚會", "1. 正常報名 1 人"),
    ("李大華", "我和一個朋友要報名迎新晚會", "2. 人數 2 人報名"),
    ("張三", "幫我和三個朋友報名迎新晚會", "3. 報名 4 人（超過 3 人限制應拒絕）"),
    ("王小明", "報名迎新晚會", "4. 重複報名同一活動（應拒絕）"),
    ("王小明", "報名程式工作坊", "5. 王小明報名第 2 個活動"),
    ("王小明", "報名籃球友誼賽", "6. 王小明報名第 3 個活動（超過上限 2 個應拒絕）"),
    ("陳美麗", "查詢籃球友誼賽名額", "7. 查詢剩餘名額"),
    ("李大華", "取消迎新晚會", "8. 取消報名釋出名額"),
    ("", "各活動報名狀況", "9. 統計查詢 (A3)")
]

print("==========================================")
print("🧪 流程測試開始 (測試數量 9 步)")
print("==========================================")

for idx, (name, cmd, desc) in enumerate(flow_steps, 1):
    print(f"\n👉 步驟 {idx} [{desc}]")
    print(f"輸入 -> 姓名: '{name}', 需求: '{cmd}'")
    reply = process_user_request(name, cmd)
    print(f"回應 ->\n{reply}")

print("\n==========================================")
print("📊 流程測試結束，顯示資料表最終狀態：")
print("==========================================")
show_tables()
````
